```python
book['Book-Title1'] = book['Book-Title'].apply(lambda x: x.replace(" ", ""))
book['Book-Author'] = book['Book-Author'].apply(lambda x: x.replace(" ", ""))
book['Publisher'] = book['Publisher'].apply(lambda x: x.replace(" ", ""))
```

```python
book['tags'] = book['tags'].apply(lambda x: x.lower())
```

```python
book = book.iloc[:,[0,1,7,9]]
```

```python
book.head()
```

```python
import nltk   # remove the like loved, loving to love only so not repeat the same word using prefix or suffix

from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()

def stem(text):
    y = []

    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)    
```

```python
book['tags'] = book['tags'].apply(stem)

book['tags']
```

```python
book.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features = 5000, stop_words='english')

vector = cv.fit_transform(book['tags'])
```

```python
# # we find the angle between the movies
# due to large data the ram is required in gb so it is not useful method

# from sklearn.metrics.pairwise import cosine_similarity

# similarity = cosine_similarity(vector)
```

```python
from sklearn.neighbors import NearestNeighbors

model = NearestNeighbors(
    n_neighbors=6,
    metric='cosine',
    algorithm='brute'
)

model.fit(vector)
```

```python
book[book['Book-Title'] == 'Classical Mythology'].index[0]
```

```python
def recommend(book_ob):
    index1 = book[book['Book-Title'] == book_ob].index[0]
    distances, indices = model.kneighbors(vector[index1:index1+1])

    for i in indices[0][1:]:
        print(book.iloc[i]['Book-Title'])
        print("Image:", book.iloc[i]['Image-URL-L'])
```

```python
recommend('Decision in Normandy')
```
